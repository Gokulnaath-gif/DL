```python
import numpy as np
import matplotlib.pyplot as plt

print("Student Name: Gokulnaath M | Roll No: 24BAD026")
print("PART A: VISUALIZATION OF ACTIVATION FUNCTIONS")

# Define mathematical functions
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def tanh(x):
    return np.tanh(x)

def relu(x):
    return np.maximum(0, x)

# Generate data
x = np.linspace(-10, 10, 100)

# Plotting
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(x, sigmoid(x), label="Sigmoid", color='blue')
plt.title("Sigmoid Activation")
plt.grid(True)

plt.subplot(1, 3, 2)
plt.plot(x, tanh(x), label="Tanh", color='green')
plt.title("Tanh Activation")
plt.grid(True)

plt.subplot(1, 3, 3)
plt.plot(x, relu(x), label="ReLU", color='red')
plt.title("ReLU Activation")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.datasets import mnist

print("Student Name: Gokulnaath M | Roll No: 24BAD026")
print("PART B: PERFORMANCE COMPARISON OF ACTIVATION FUNCTIONS")

# Load and preprocess data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0

def build_model(activation_func):
    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(128, activation=activation_func),
        Dense(64, activation=activation_func),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

activations = ['sigmoid', 'tanh', 'relu']
history_dict = {}

for act in activations:
    print(f"\nTraining model with {act.upper()} activation...")
    model = build_model(act)
    history = model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test), verbose=1)
    history_dict[act] = history

# Plot Accuracy Comparison
plt.figure(figsize=(10, 5))
for act in activations:
    plt.plot(history_dict[act].history['val_accuracy'], label=f"{act} Val Accuracy")
plt.title('Validation Accuracy Comparison across Activations')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Student Name: Gokulnaath M | Roll No: 24BAD026")
print("PART C: COMPARISON OF OPTIMIZATION ALGORITHMS")

def build_model_for_optimizer(opt):
    model = Sequential([
        Flatten(input_shape=(28, 28)),
        Dense(128, activation='relu'),
        Dense(64, activation='relu'),
        Dense(10, activation='softmax')
    ])
    model.compile(optimizer=opt, loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

# Define optimizers
optimizers = {
    'SGD': tf.keras.optimizers.SGD(learning_rate=0.01),
    'Momentum': tf.keras.optimizers.SGD(learning_rate=0.01, momentum=0.9),
    'RMSProp': tf.keras.optimizers.RMSprop(learning_rate=0.001),
    'Adam': tf.keras.optimizers.Adam(learning_rate=0.001)
}

opt_history_dict = {}

for opt_name, opt in optimizers.items():
    print(f"\nTraining model with {opt_name} optimizer...")
    model = build_model_for_optimizer(opt)
    history = model.fit(x_train, y_train, epochs=5, validation_data=(x_test, y_test), verbose=1)
    opt_history_dict[opt_name] = history

# Plot Loss vs Epoch
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
for opt_name in optimizers.keys():
    plt.plot(opt_history_dict[opt_name].history['loss'], label=f"{opt_name} Train Loss")
plt.title('Training Loss vs Epochs')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Plot Accuracy vs Epoch
plt.subplot(1, 2, 2)
for opt_name in optimizers.keys():
    plt.plot(opt_history_dict[opt_name].history['val_accuracy'], label=f"{opt_name} Val Accuracy")
plt.title('Validation Accuracy vs Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
